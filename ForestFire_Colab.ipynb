## Step 1 — Clone Repository


````python
import os
REPO_DIR = '/content/forest-fire-intelligence'
if os.path.exists(REPO_DIR):
    import shutil; shutil.rmtree(REPO_DIR)
os.system(f'git clone https://github.com/rohithas-dev/india-forest-fire-intelligence {REPO_DIR}')
os.chdir(REPO_DIR)
print('✅ Cloned:', os.getcwd())
os.system('ls -la')
````

## Step 2 — Install All Dependencies

> This takes about 2 minutes the first time.


````python
import subprocess, sys
subprocess.run([sys.executable, '-m', 'pip', 'install', '-q', '-r', 'requirements.txt'], check=False)
subprocess.run([sys.executable, '-m', 'pip', 'install', '-q', 'pyngrok', 'streamlit-folium'], check=False)
print('✅ All packages installed!')
````

## Step 3 — Initialize Database with 1,200 Demo Fire Events


````python
import os, subprocess, sys
os.chdir('/content/forest-fire-intelligence')
result = subprocess.run([sys.executable, 'scripts/initialize_database.py'], capture_output=False)
print('✅ Database ready!' if result.returncode == 0 else '⚠️ Check output above')
````

## Step 4 — Configure ngrok (Free Public URL)

**Get your free ngrok token (1 minute):**
1. Sign up at 👉 **[https://ngrok.com](https://ngrok.com)**
2. After login, go to 👉 **[https://dashboard.ngrok.com/get-started/your-authtoken](https://dashboard.ngrok.com/get-started/your-authtoken)**
3. Copy your Authtoken
4. Paste it below, replacing `YOUR_TOKEN_HERE`


````python
from pyngrok import ngrok, conf

# ✏️  PASTE YOUR NGROK TOKEN BELOW
NGROK_TOKEN = 'YOUR_TOKEN_HERE'
# ============================================================

ngrok.kill()
conf.get_default().auth_token = NGROK_TOKEN
print('✅ ngrok configured!')
````

## Step 5 — Launch the App 🚀

Run this cell. After ~10 seconds your **live public URL** will appear below.

```
🔥 YOUR APP IS LIVE AT: https://xxxx-xx-xx.ngrok-free.app
```


````python
import subprocess, time, re, os, sys
from pyngrok import ngrok

os.chdir('/content/forest-fire-intelligence')

# Stop any existing app
os.system('pkill -f streamlit 2>/dev/null')
time.sleep(2)

# Start Streamlit in the background
proc = subprocess.Popen(
    [sys.executable, '-m', 'streamlit', 'run', 'app.py',
     '--server.port', '8501',
     '--server.headless', 'true',
     '--server.enableCORS', 'false',
     '--server.enableXsrfProtection', 'false',
     '--browser.gatherUsageStats', 'false'],
    stdout=subprocess.PIPE, stderr=subprocess.STDOUT
)

print('⏳ Starting Streamlit (10 seconds)...')
time.sleep(10)

# Open public tunnel
public_url = ngrok.connect(8501)
url_str = str(public_url)
m = re.search(r'https://[\w.-]+\.ngrok[\w-]*\.app|https://[\w.-]+\.ngrok\.io', url_str)
clean_url = m.group(0) if m else url_str.split('"')[1] if '"' in url_str else url_str

print()
print('=' * 60)
print('🔥  YOUR APP IS LIVE AT:')
print('   ' + clean_url)
print('=' * 60)
print()
print('👆 Click the link above to open your Forest Fire Intelligence app!')
print('⚠️  Keep this Colab tab open — closing it stops the app.')
print('ℹ️  To stop: run the STOP cell at the bottom.')
````

````python
import plotly.express as px
fire_df = load_fire_events_df(limit=2000)
if not fire_df.empty:
    fig = px.scatter_geo(
        fire_df, lat='latitude', lon='longitude',
        color='fire_radiative_power', size='brightness',
        color_continuous_scale='Inferno', projection='natural earth',
        title='🔥 Demo Fire Events — Forest Fire Intelligence',
        hover_data=['country', 'season', 'confidence']
    )
    fig.update_layout(paper_bgcolor='#0A0E1A', font_color='#E8EAF0', height=500)
    fig.show()
else:
    print('No fire data. Run Step 3 to seed the database.')
````
